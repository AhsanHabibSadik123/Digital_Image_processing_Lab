# 11. Write a python program to apply log and power-law (gamma) transformations.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("image.jpeg", cv2.IMREAD_GRAYSCALE)
```

```python
# Log transform
c = 255 / np.log(1 + 255)
log_img = c * np.log(1 + img)
log_img = np.uint8(log_img)
```

```python
# gamma transform
gamma = .5
c = 255 / (255 ** gamma)
gamma_img = c * img ** gamma
```

```python
plt.figure(figsize=(20, 10))
plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("original image")

plt.subplot(1, 3, 2)
plt.imshow(log_img, cmap="gray")
plt.title("log transformed image")

plt.subplot(1, 3, 3)
plt.imshow(gamma_img, cmap="gray")
plt.title("gamma transformed image")
```
