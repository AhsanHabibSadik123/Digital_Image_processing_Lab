# 12. Write a python program to implement spatial domain smoothing/averaging filters.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("image.jpeg", cv2.IMREAD_GRAYSCALE)
```

```python
height, width = img.shape
```

```python
smooth_img = np.zeros_like(img)

for i in range(1, height - 2):
    for j in range(1, width - 2):
        matrix = img[i-1:i+3, j-1:j+3]
        avg = np.sum(matrix) / 16
        smooth_img[i][j] = np.int8(avg)
```

```python
plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(smooth_img, cmap="gray")
plt.title("Smooth image")
plt.axis("off")
```
