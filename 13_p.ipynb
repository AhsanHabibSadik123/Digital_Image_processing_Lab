# 13. Write a python program to implement non-linear filters (Median, Max, Min filters) for noise reduction.

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
img = cv2.imread("image.jpeg", cv2.IMREAD_GRAYSCALE)
```

```python
# Max Filter
max_filter_img = np.zeros_like(img)
height, width = img.shape
for i in range(1, height-1):
    for j in range(1, width-1):
        matrix = img[i-1:i+2, j-1:j+2]
        max_filter_img[i][j] = np.max(matrix)
```

```python
# Min Filter
min_filter_img = np.zeros_like(img)
height, width = img.shape
for i in range(1, height-1):
    for j in range(1, width-1):
        matrix = img[i-1:i+2, j-1:j+2]
        min_filter_img[i][j] = np.min(matrix)
```

```python
# Median Filter
median_filter_img = np.zeros_like(img)
height, width = img.shape
for i in range(1, height-1):
    for j in range(1, width-1):
        matrix = img[i-1:i+2, j-1:j+2]
        median_filter_img[i][j] = np.median(matrix)
```

```python
plt.subplot(2, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(max_filter_img, cmap="gray")
plt.title("Max filter image")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(min_filter_img, cmap="gray")
plt.title("Min filter image")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(median_filter_img, cmap="gray")
plt.title("Median filter image")
plt.axis("off")
```
