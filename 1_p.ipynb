# 1. Write a python program to Read, display, and convert image types (RGB to Grayscale, Binary).

```python
from PIL import Image
import matplotlib.pyplot as plt
import cv2
```

```python
image_cv = cv2.imread("image.jpeg")
image_cv = cv2.cvtColor(image_cv, cv2.COLOR_BGR2RGB)
plt.imshow(image_cv)
```

```python
image_pil = Image.open("image.jpeg")
plt.imshow(image_pil)
```

```python
# RGB to Gray
weight, height = image_pil.size
gray_img = Image.new("L", (weight, height))
image_pil.size, gray_img.size
```

```python
rgb_image = image_pil.convert("RGB")
for j in range(height):
    for i in range(weight):
        r, g, b = rgb_image.getpixel((i, j))
        gray = 0.299 * r + 0.587 * g + 0.114 * b
        gray_img.putpixel((i, j), int(gray))

plt.imshow(gray_img,cmap='gray')
```

```python
binary_image = Image.new("L", (weight, height))
threshold = 127
for j in range(height):
    for i in range(weight):
        gray = gray_img.getpixel((i, j))
        if gray > threshold:
            binary_image.putpixel((i, j), int(255))
        else:
            binary_image.putpixel((i, j), int(0))

plt.imshow(binary_image,cmap='gray')
```
