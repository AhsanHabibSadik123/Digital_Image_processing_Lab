# 2. Write a python program to analyze the effects of changing spatial and intensity resolution.

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image = cv2.imread("image.jpeg", cv2.IMREAD_GRAYSCALE)
# image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
image.shape
```

```python
# Spatial Resulation

spatial_128 = cv2.resize(image, (128, 128))
spatial_64 = cv2.resize(image, (64, 64))
spatial_32 = cv2.resize(image, (32, 32))
```

```python
intensity_256 = image
intensity_16 = np.floor(image / 64) * 64
intensity_8 = np.floor(image / 8) * 8
intensity_4 = np.floor(image / 4) * 4
```

```python
# -----------------------------
# Display Spatial Resolution
# -----------------------------

plt.figure(figsize=(10, 6))

plt.subplot(2, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(spatial_128, cmap="gray")
plt.title("128 x 128")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(spatial_64, cmap="gray")
plt.title("64 x 64")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(spatial_32, cmap="gray")
plt.title("32 x 32")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# -----------------------------
# Display Intensity Resolution
# -----------------------------

plt.figure(figsize=(10, 6))

plt.subplot(2, 2, 1)
plt.imshow(intensity_256, cmap="gray")
plt.title("256 Levels")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(intensity_16, cmap="gray")
plt.title("16 Levels")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(intensity_8, cmap="gray")
plt.title("8 Levels")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(intensity_4, cmap="gray")
plt.title("4 Levels")
plt.axis("off")

plt.tight_layout()
plt.show()
```
