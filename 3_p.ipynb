# 3. Write a python program to perform geometric transformations (Resizing/Zooming, Rotation, Translation).

```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("image.jpeg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img)
```

```python
height, width = img.shape[:2]
```

```python
# Resizing / Zooming
resized_img = cv2.resize(img, (int(height * .5), int(width * .5)))
plt.imshow(resized_img)
```

```python
# Rotation
rotational_matrix = cv2.getRotationMatrix2D(
    (width // 2, height // 2),
    45,
    1
)
rotate_img = cv2.warpAffine(
    img,
    rotational_matrix,
    (width, height)
)
plt.imshow(rotate_img)
```
