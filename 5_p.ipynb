# 5. Write a program to convert color image to gray scale image, Color image into Binary Image, Separate color image into three separate R, G, B planes, combine three separate R, G, B planes to create color image.

```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("image.jpeg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
gray_image = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(gray_image, cmap="gray")
```

```python
r = img[:, :, 0]
g = img[:, :, 1]
b = img[:, :, 2]
```

```python
plt.figure(figsize=(10, 10))

plt.subplot(2, 2, 1)
plt.imshow(img)
plt.title("Original Image")

plt.subplot(2, 2, 2)
plt.imshow(r, cmap='gray')
plt.title("Red Channel")

plt.subplot(2, 2, 3)
plt.imshow(g, cmap="gray")
plt.title("Green Channel")

plt.subplot(2, 2, 4)
plt.imshow(b, cmap='gray')
plt.title("Blue Channel")

```

```python
original_img = cv2.merge([r, g, b])
plt.imshow(original_img)
```
