# 6. Write a python program to perform arithmetic (addition, subtraction) and logical (AND, OR, NOT) operations.

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
white_img = np.ones((148, 148), dtype=np.uint8) * 255
black_img = np.zeros((148, 148), dtype=np.uint8)
white_img.shape, black_img.shape
```

```python
add_img = cv2.add(white_img, black_img)
cv2.imshow("Add image", add_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
sub_img = cv2.subtract(white_img, black_img)
cv2.imshow("subtract image", sub_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
and_img = cv2.bitwise_and(white_img, black_img)
cv2.imshow("And image", and_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
or_img = cv2.bitwise_or(white_img, black_img)
cv2.imshow("Or image", or_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
not_img = cv2.bitwise_not(black_img)
cv2.imshow("Not image", not_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
