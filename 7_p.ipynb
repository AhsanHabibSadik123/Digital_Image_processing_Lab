# 7. Write a python program to implement image negative and thresholding operations.

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img = cv2.imread("image.jpeg", cv2.IMREAD_GRAYSCALE)
img.shape
```

```python
neg_img = 255 - img
plt.imshow(neg_img, cmap="gray")
```

```python
width, height = img.shape
```

```python
threshold = 127
thres_img = np.zeros((img.shape))
for i in range(height):
    for j in range(width):
        gray = img[j][i]
        if gray > 127:
            thres_img[j][i] = 255
            
plt.imshow(thres_img, cmap="gray")
```
