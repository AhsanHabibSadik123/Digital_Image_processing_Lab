# 8. Write a python program to apply gray-level slicing (with and without background).

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("image.jpeg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
```

```python
low_thres = 100
high_thres = 180
```

```python
with_bg_img = np.where((img >= low_thres) & (img <= high_thres), 255, img)
without_bg_img = np.where((img >= low_thres) & (img <= high_thres), 255, 0)
```

```python
plt.figure(figsize=(10, 10))
plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Original image")

plt.subplot(1, 3, 2)
plt.imshow(with_bg_img, cmap="gray")
plt.title("With background")

plt.subplot(1, 3, 3)
plt.imshow(without_bg_img, cmap="gray")
plt.title("With out bg")
```
