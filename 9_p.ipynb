# 9. Write a python program to perform contrast stretching and bit-plane slicing.

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
img = cv2.imread("image.jpeg", cv2.IMREAD_GRAYSCALE)
```

```python
min_value = np.min(img)
max_value = np.max(img)
contrast_stretching_img = ((img - min_value) / (max_value - min_value) * 255).astype("uint8")
```

```python
plt.subplot(1, 2, 1)
plt.imshow(contrast_stretching_img, cmap="gray")
plt.title("contrast stretched img")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(img, cmap="gray")
plt.title("original img")
plt.axis("off")
```

```python
bit_plane = []
for i in range(8):
    plane = ((img >> i) & 1) * 255
    bit_plane.append(plane)
```

```python
plt.figure(figsize=(12, 8))
for i in range(8):
    plt.subplot(2, 4, i + 1)
    plt.imshow(bit_plane[i], cmap="gray")
    plt.title(f"Bit plane {i + 1}")
    plt.axis('off')
```
